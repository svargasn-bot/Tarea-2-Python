## Matriz de correlacion

```python
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include='number')
corr = numeric_df.corr()

plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Correlación')
plt.xticks(range(len(corr)), corr.columns, rotation=45, ha='right')
plt.yticks(range(len(corr)), corr.columns)
for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(j, i, f"{corr.iat[i, j]:.2f}", ha='center', va='center', color='black')
plt.title('Matriz de correlación')
plt.tight_layout()
```

### Correlación Positiva Moderada: Edad vs. Años de Experiencia:
Es la correlación más fuerte de todo el gráfico (sacando la diagonal). Tiene todo el sentido del mundo: a mayor edad, por lo general, la persona ha tenido más tiempo para acumular años de experiencia laboral.

### Ausencia de Correlación: Las variables de desempeño son independientes: 
Las notas que ponen los clientes y su nivel de satisfacción no dependen de la edad del freelancer, ni de cuánta experiencia tenga, ni de cuánto cobre por hora. Un freelancer joven, con poca experiencia y barato puede tener un rating tan perfecto de 5 estrellas como un senior veterano y caro. El talento y el buen servicio son independientes de la demografía.
